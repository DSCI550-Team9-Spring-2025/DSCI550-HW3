# Spike Map Data Preprocessing
This notebook processes Data/v2_final.tsv into appropriate description JSON chunks for D3 Spike Map visualization.

```python
import numpy as np
import pandas as pd
import json
import os
```

### Load TSV

```python
tsvpath = os.path.join("..", "Data", "v2_final.tsv")
df = pd.read_csv(tsvpath, sep='\t')
```

```python
df = df.set_index('index')
df.head()
```

### Generate JSONs

```python
sightings_data = df[["city", "state", "city_longitude", "city_latitude"]].dropna().to_dict(orient="records")
with open(os.path.join("react-ui", "public", "SightingMap", "us.json"), 'w') as f:
    json.dump(sightings_data, f, indent=2, ensure_ascii=False)
```
