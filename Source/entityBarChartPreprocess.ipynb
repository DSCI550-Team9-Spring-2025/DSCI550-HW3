# Entity Bar Chart Preprocessing
This notebook processes Data/v2_final.tsv into appropriate description JSON chunks for D3 sortable bar chart visualization.

```python
import numpy as np
import pandas as pd
import json
import os
```

### Load TSV

```python
tsvpath = os.path.join("..", "Data", "v2_final.tsv")
df = pd.read_csv(tsvpath, sep='\t')
```

```python
df = df.set_index('index')
df.head()
```

### Generate JSONs

```python
from collections import defaultdict
entities = defaultdict(int)
```

```python
for row in df.spacy_entities:
    for entry in row.split('\n'):
        try:
            entry = entry.split(",")
            ent = entry[len(entry) - 1].strip()
            entities[ent] += 1
        except IndexError:
            continue
```

```python
try:
    del entities['']
except KeyError:
    ...
entities
```

```python
with open(os.path.join("react-ui", "public", "EntityBarChart", "spacy_entities.json"), 'w') as f:
    json.dump(entities, f, indent=2, ensure_ascii=False)
```
